```python
import pandas as pd

url_dados_spotify = '/content/high_popularity_spotify_data - high_popularity_spotify_data.csv.csv'

df_spotify = pd.read_csv(url_dados_spotify)

display(df_spotify[['track_name', 'track_artist']].head(50))
```

```python
#identificar o operador de separação
df_spotify['artistas_separados'] = df_spotify['track_artist'].str.split(',')
display(df_spotify['artistas_separados'])

# vamos explodir esses dados para mostrar individualmente
df_explodido = df_spotify.explode('artistas_separados')

#vamos limpar esse dado, para remover os espaços no começo e no fim
df_explodido['artistas_separados'] = df_explodido['artistas_separados'].str.strip()

display(df_explodido)

display(df_explodido[['artistas_separados', 'track_name']].head(50))
```

```python
# vamos encontrar um panorama geral
media_popularidade = df_explodido['track_popularity'].mean()
print(media_popularidade)

mediana_popularidade = df_explodido['track_popularity'].median()
print(mediana_popularidade)

moda_popularidade = df_explodido['track_album_name'].mode()[0]
print(moda_popularidade)

# para nossa amplitudade, vamos aproveitar e descobrir, qual a musica mais poupular
# e a menos popular
max_popularidade = df_explodido['track_popularity'].max()
min_popularidade = df_explodido['track_popularity'].min()
amplitude_popularidade = max_popularidade - min_popularidade
print(f''' {max_popularidade}
      {min_popularidade}
      {amplitude_popularidade}
      ''')
indice_mais_popular = df_explodido['track_popularity'].idxmax()

musica_mais_popular = df_explodido.loc[indice_mais_popular, 'track_name']
print(musica_mais_popular)


musica_menos_popular = df_explodido.loc[indice_menos_popular, 'track_name']
print(musica_menos_popular)
```

```python
# vamos encontrar um panorama geral
media_popularidade = df_explodido['track_popularity'].mean()
print(media_popularidade)

mediana_popularidade = df_explodido['track_popularity'].median()
print(mediana_popularidade)

moda_popularidade = df_explodido['track_album_name'].mode()[0]
print(moda_popularidade)

# para nossa amplitudade, vamos aproveitar e descobrir, qual a musica mais poupular
# e a menos popular
max_popularidade = df_explodido['track_popularity'].max()
min_popularidade = df_explodido['track_popularity'].min()
amplitude_popularidade = max_popularidade - min_popularidade
print(f''' {max_popularidade}
      {min_popularidade}
      {amplitude_popularidade}
      ''')
indice_mais_popular = df_explodido['track_popularity'].idxmax()

musica_mais_popular = df_explodido.loc[indice_mais_popular, 'track_name']
print(musica_mais_popular)

indice_menos_popular = df_explodido['track_popularity'].idxmin()
musica_menos_popular = df_explodido.loc[indice_menos_popular, 'track_name']
print(musica_menos_popular)
```

```python
import matplotlib.pyplot as plt

# para criar uma amostragem, ou seja, pegar aleatoriamente
amostra = df_explodido.sample(100, random_state=42)
#grafico de dispersao
plt.scatter(amostra['energy'], amostra ['danceability'])
plt.xlabel('energy')
plt.ylabel('danceability')

plt.show()

display(amostra[['energy', 'danceability', 'track_name']])
```

```python
import plotly.express as px

# Criar o gráfico de dispersão interativo com Plotly Express
fig = px.scatter(amostra, x='energy', y='danceability',
                 title='Relação entre Energia e Danceabilidade das Músicas (Amostra)')

# Exibir o gráfico
fig.show()
```
