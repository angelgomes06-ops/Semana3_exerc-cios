```python
# queremos ler a base de dados e interpretar seus campos
# (pandas, display(), info(), describe())

import pandas as pd
df_logs = pd.read_csv('/content/logs_seguranca_aula.csv')

display(df_logs.head(10))


# SEMPRE QUE TEMOS UM "timestamp" DEVEMOS VERIFICAR SE ELE É UM
df_logs['timestamp'] = pd.to_datetime(df_logs['timestamp'])
df_logs.info()
df_logs.describe()
# conversao feita para trabalhar com data/hora (campo veio como texto e transformou data/hora)

```

```python
# descobrir logins realizados fora do horario comercial
# seg - sex - 08 - 17
# o python pode identificar dentro de um timestamp valores ramificados
df_logs['hora'] = df_logs ['timestamp'].dt.hour
df_logs['minutos'] = df_logs['timestamp'].dt.minute
display(df_logs)
df_logs['dia_da_semana'] = df_logs ['timestamp'].dt.dayofweek

# criar os filtros
#fora do horario comercial hora 0 a 7 - 17 até 23
display(df_logs[df_logs['hora'] !=  8])

# vamos exibir os dados que nao sao resource financeiro ou system

# or -----> dentro de if, for ou while
# | ----> operador OR
# and ---> dentro de if, for ou while
# & ---> operador de E ou AND

display(df_logs[(df_logs['resource'] != 'financeiro')  &
                (df_logs['resource'] != 'system')])

# or - | é quando pelo menos uma das solucoes é verdade
# and - & quando todas as solucoes sao verdade

#filtrar colunas diferentes
# somente falhas na sexta- feira

display(df_logs[(df_logs['status'] == 'FAILED') &
                (df_logs['dia_da_semana'] == 6)])

#encontrar informacoes que iniciam ou finalizam com
# starswith - inicia com
# endswith - finaliza com
# mostrar tudo que termina com .csv
display (df_logs[df_logs['resource'].str.endswith('.csv', na=False)])
```
