```python
import pandas as pd
import numpy as np # o numby são as funções matematicas para python
import json

# 1. Gerando o arquivo CSV e Excel (example_data)
df_example = pd.DataFrame({
    'Name': ['Ana', 'Bruno', 'Carlos', 'Diana', 'Eduardo'],
    'Age': [22, 28, 26, 24, 30],
    'Score': [85, 90, 78, 92, 88]
})
df_example.to_csv('example_data.csv', index=False)
df_example.to_excel('example_data.xlsx', index=False)

# 2. Gerando o arquivo JSON aninhado (example_data.json)
json_data = {
    "employees": [
        {"id": 1, "department": "Vendas", "info": {"Name": "Ana", "Role": "Analista"}},
        {"id": 2, "department": "TI", "info": {"Name": "Bruno", "Role": "Gerente"}}
    ]
}
with open('example_data.json', 'w') as f:
    json.dump(json_data, f)

# 3. Gerando os dados para manipulação avançada (sample_data.csv)
df_sample = pd.DataFrame({
    'age': [25, np.nan, 30, 28, np.nan, 35], # NaN representa dados ausentes
    'salary': [5000, 6000, 5500, 7000, 6500, 8000],
    'department': ['TI', 'TI', 'RH', 'RH', 'Vendas', 'Vendas']
})
df_sample.to_csv('sample_data.csv', index=False)

print("Bases de dados geradas com sucesso no Colab!")
```

```python
df_data = pd.read_csv('sample_data.csv')

#vamos usar o python para ver valores ausentes por coluna
print('Valores ausentes: \n\n', df_data.isnull().sum())

# vamos substituir as idades vazias por médias
df_data['age'] = df_data ['age'].fillna(df_data['age'].mean())
# para substituir um valor VAZIO chama a base com o campo que vc quer
# a base recebe ela mesm com o método .fillna << o fillna filtra valores vazios
# e troca por outra coisa >> .fillna(o que quer que entre no lugar exemplo média)

display(df_data)

# vamos exibir a média salarial por dpto
# vamos usar um agrupamento de categorias com python

df_data_salario_por_departamento = df_data.groupby('department')['salary'].mean
display(df_data_salario_por_departamento)
```
