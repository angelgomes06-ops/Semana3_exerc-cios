```python
# para um melhor relatório, uma tecnica que podemos fazer é unir base de dados para um visual unico

# queremos unir a base vendas com clientes

import pandas as pd

# vamos chamar as tabelas que queremos dar merge (mesclagem/unir)
df_vendas = '/content/dados/vendas.csv'
df_clientes = '/content/dados/clientes.csv'

df_vendas = pd.read_csv(df_vendas)
df_clientes = pd.read_csv(df_clientes)

df_vendas_clientes = pd.merge(df_vendas,
                              df_clientes,
                              left_on='Cliente',
                              right_on='Id Cliente',
                              how= 'inner')

#df_nome_tabela = pd.merge(tabela da esquerda, tabela da direita,
#                           left on= 'nome da tabela da esquerda'
#                          right on='nome da tabela da esquerda')

display(df_vendas_clientes)
```

```python
# queremos em uma unica tabela o quanto o nosso cliente gastou em sua compra


# ponto 1 - descobrir onde esta o valor do produto
df_produtos = '/content/dados/produtos.csv'
df_produtos = pd.read_csv(df_produtos)

# DF_VENDAS_CLIENTES [TAB]

# ponto 2 - mesclar a tabela df_vendas_clientes com a tabela que tem o valor
df_vendas_clientes_produtos = pd.merge(
                                       df_vendas_clientes,
                                      df_produtos,
                                      left_on= 'Produto',
                                      right_on= 'Produto',
                                      how= 'inner')
# ponto 3- criar um novo campo, para mostrar o total gasto da compra
df_vendas_clientes_produtos['Total'] = df_vendas_clientes_produtos['Quantidade'] * df_vendas_clientes_produtos ['Valor']
# ponto 4 - display
display(df_vendas_clientes_produtos)

#PONTO 5 - salvar o csv no computador
# para salvar o seu df no computador
df_vendas_clientes_produtos.to_csv('Dados completos.csv', index=False)
```
