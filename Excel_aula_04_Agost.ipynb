```python
import requests
import json

endpoint_sheets = 'https://script.google.com/macros/s/AKfycbxpYJlotxJKX36DeniV9eAAabJ-Fhta3NeO2aNxEgijSpdSaCZ5ExabbzvLrhCBGJhz4w/exec'

#post - postar - colocar
#get - pegar - coleto uma informação de algum lugar

def enviar_msg(nome, email):
  response = requests.post(
      #onde
      endpoint_sheets,
      #o que
      json={"nome": nome, "email": email}
  )
  #para enviar uma msg do python para o sheets, devemos
  #1 - criar a função
  #2 - chamar o metodo request.post (onde, o que)

  #se quiser ver a mensagem enviada
  print(response.text)
```

```python
nome = input ('informe o nome desejado\n')
email = input ('informe o email desejado\n')

enviar_msg(nome, email)
```
